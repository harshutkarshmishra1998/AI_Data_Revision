```python
# 1. Given a list of employee dictionaries, group employees by department.
# Write your solution below
employees = [
    {"id": 101, "name": "Alice", "department": "IT", "salary": 75000},
    {"id": 102, "name": "Bob", "department": "HR", "salary": 55000},
    {"id": 103, "name": "Charlie", "department": "IT", "salary": 80000},
    {"id": 104, "name": "David", "department": "Finance", "salary": 65000},
    {"id": 105, "name": "Emma", "department": "HR", "salary": 60000},
    {"id": 106, "name": "Frank", "department": "Finance", "salary": 70000},
    {"id": 107, "name": "Grace", "department": "IT", "salary": 85000},
    {"id": 108, "name": "Henry", "department": "Marketing", "salary": 58000},
    {"id": 109, "name": "Ivy", "department": "Marketing", "salary": 62000},
    {"id": 110, "name": "Jack", "department": "HR", "salary": 57000}
]

# for emp in employees:
#     for k, v in emp.items():
#         print(k, v)

grouped_emp = {}

for emp in employees:
    dept = emp["department"]
    if dept not in grouped_emp:
        grouped_emp[dept] = []
    grouped_emp[dept].append(emp["name"])

print(grouped_emp)
```
