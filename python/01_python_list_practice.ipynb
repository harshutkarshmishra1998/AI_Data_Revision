```python
# 1. Find the second-largest unique element in a list. Example: [10, 20, 4, 20, 15] -> 15
# Write your solution below

list_1 = [-1, 10, 20, 4, 20, 15, 18, 50]

# unique_list = list(set(list_1))
# print(unique_list)

high_1 = list_1[0]
high_2 = list_1[0]

for item in list_1:
    if item > high_1: # Get the highest number
        high_2 = high_1
        high_1 = item
    if high_1 > item > high_2: # Get the second highest
        high_2 = item

print(high_1, high_2)
```
