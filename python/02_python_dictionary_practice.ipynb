```python
# 1. Count the frequency of each character in a string. Example: 'programming' -> character frequency dictionary.
# Write your solution below

word = "programming"

word_list = list(word)
print(word_list)

word_dict ={}

for items in word_list:
    if items in word_dict:
        word_dict[items] += 1
    else:
        word_dict[items] = 1

print(word_dict)
```

```python
# 2. Count the frequency of each word in a sentence. Example: 'AI is great and AI is useful' -> {'AI': 2, 'is': 2, 'great': 1, 'and': 1, 'useful': 1}
# Write your solution below

sent = "AI is great and AI is useful"

sent_list = sent.split()
print(sent_list)

sent_dict = {}

for items in sent_list:
    if items in sent_dict:
        sent_dict[items] += 1
    else:
        sent_dict[items] = 1

print(sent_dict)
```

```python
# 3. Find the key with the maximum value in a dictionary. Example: {'A': 20, 'B': 50, 'C': 30} -> 'B'
# Write your solution below

dict_1 = {'A': 20, 'B': 50, 'C': 30}

high_key = ""
high_value = float('-inf')

for k,v in dict_1.items():
    if v > high_value:
        high_key, high_value = k, v

print (high_key, high_value)
```
