```python
import numpy as np
import pandas as pd

# Creating a sample dataset with missing values
data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Department": ["HR", "IT", None, "Finance", "HR"],
    "Salary": [50000, np.nan, 65000, np.nan, 72000],
    "Rating": [4.5, 3.8, np.nan, np.nan, 4.9],
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

print(df.isnull())

# Step 1: Count missing values per column
missing_counts = df.isnull().sum()
print("\nMissing values count per column:")
print(missing_counts)

# Step 2: Calculate percentage of missing values to assess severity
missing_percentage = (df.isnull().sum() / len(df)) * 100
print("\nMissing values percentage:")
print(round(missing_percentage, 2))
```

```python
import numpy as np
import pandas as pd

# Creating a dataset with an extreme outlier
data = {
    "Transaction_ID": [1, 2, 3, 4, 5, 6, 7, 8],
    "Order_Amount": [12000000, 150, 135, 142, 128, 145, 138, 950000000]  # 95,000 is an outlier
}

df = pd.DataFrame(data)

# Step 1: Calculate mean and standard deviation of the column
mean_val = df["Order_Amount"].mean()
std_val = df["Order_Amount"].std()

print(f"Mean: {mean_val:.2f}, Standard Deviation: {std_val:.2f}")

# Step 2: Compute the Z-score for each row
df["Z_Score"] = (df["Order_Amount"] - mean_val) / std_val

print("\nDataFrame with Z-Scores:")
print(df)

# Step 3: Identify outliers where absolute Z-score is greater than 3
outliers = df[df["Z_Score"].abs() > 3]
print("\nIdentified Outliers (|Z| > 3):")
print(outliers)
```

```python
import numpy as np
import pandas as pd

# Larger dataset to allow the mean and std to stabilize
np.random.seed(42)
normal_data = np.random.normal(loc=100, scale=10, size=100)
# Injecting an extreme outlier
data_with_outlier = np.append(normal_data, [500])

df = pd.DataFrame({"Value": data_with_outlier})

# Calculate mean and standard deviation
mean_val = df["Value"].mean()
std_val = df["Value"].std()

# Compute Z-score
df["Z_Score"] = (df["Value"] - mean_val) / std_val

# Filter outliers (|Z| > 3)
outliers = df[df["Z_Score"].abs() > 3]

print(f"Mean: {mean_val:.2f}, Std: {std_val:.2f}")
print("\nIdentified Outliers using Z-score > 3:")
print(outliers)
```

```python
import pandas as pd
import numpy as np

# Sample dataset with various flaws (constant col, high missing, high correlation)
data = {
    "Transaction_ID": [101, 102, 103, 104, 105], # High cardinality ID (Drop)
    "Constant_Col": [5, 5, 5, 5, 5],             # Zero variance (Drop)
    "Missing_Col": [np.nan, np.nan, np.nan, 10, 20], # High missing values (Drop)
    "Square_Feet": [1000, 1500, 2000, 2500, 3000],
    "Price_In_Lakhs": [50, 75, 100, 125, 150],    # Highly correlated with Square_Feet
    "Target": [0, 1, 0, 1, 1]
}
df = pd.DataFrame(data)

print("Original Data Shape:", df.shape)

# --- Step 1: Drop columns with > 50% Missing Values ---
missing_percentage = df.isnull().mean()
cols_to_drop_missing = missing_percentage[missing_percentage > 0.5].index.tolist()
df = df.drop(columns=cols_to_drop_missing)
print(f"Dropped due to high missing values: {cols_to_drop_missing}")

# --- Step 2: Drop Constant / Low Variance Columns ---
# Check standard deviation or unique counts for zero variance
variance = df.var(numeric_only=True)
cols_to_drop_variance = variance[variance == 0].index.tolist()
df = df.drop(columns=cols_to_drop_variance)
print(f"Dropped due to zero variance: {cols_to_drop_variance}")

# --- Step 3: Drop Highly Correlated Features (Multicollinearity) ---
# Calculate correlation matrix
corr_matrix = df.corr(numeric_only=True).abs()

print(corr_matrix)

# Find pairs with correlation > 0.90
upper_tri = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
to_drop_corr = [column for column in upper_tri.columns if any(upper_tri[column] > 0.90)]
df = df.drop(columns=to_drop_corr)
print(f"Dropped due to high collinearity (>0.90): {to_drop_corr}")

print("\nCleaned Data Shape after Analysis & Pruning:", df.shape)
print(df.head())
```

```python
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Step 1: Generate a high-dimensional dataset (1,000 rows, 20 highly correlated features)
np.random.seed(42)
n_samples = 1000

# Create two hidden underlying patterns (latent factors)
factor_1 = np.random.normal(0, 1, n_samples)
factor_2 = np.random.normal(0, 1, n_samples)

# Generate 20 features tightly correlated with these factors
data = {}
for i in range(10):
    data[f"Metric_GroupA_{i}"] = factor_1 + np.random.normal(0, 0.05, n_samples)
for i in range(10):
    data[f"Metric_GroupB_{i}"] = factor_2 + np.random.normal(0, 0.05, n_samples)

df_high_dim = pd.DataFrame(data)
print(f"Original Dataset Shape: {df_high_dim.shape}")

# Step 2: Standardize the features (Mandatory for PCA so high-magnitude columns don't dominate)
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df_high_dim)

# Step 3: Automatically retain enough components to preserve 95% of total dataset variance
pca_optimal = PCA(n_components=None)
reduced_data = pca_optimal.fit_transform(scaled_data)

print(f"\nOriginal feature count: {df_high_dim.shape[1]}")
print(f"Optimal feature count needed for 95% variance: {reduced_data.shape[1]}")
print(f"Exact variance retained: {pca_optimal.explained_variance_ratio_.sum() * 100:.2f}%")

# Step 4: Inspect individual explained variance ratios of the selected components
print("\nExplained Variance Ratio per Selected Component:")
print(pca_optimal.explained_variance_ratio_.round(4))
```

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Sample data with 3 correlated features
data = {
    "Study_Hours": [2, 3, 5, 8, 10],
    "Attendance_Pct": [50, 60, 75, 90, 95],
    "Sleep_Hours": [7, 6, 8, 5, 6]
}
df = pd.DataFrame(data)

print(df.corr())

# Step 1: Standardize data first (mandatory for meaningful PCA/covariance analysis)
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)
df_scaled = pd.DataFrame(scaled_data, columns=df.columns)

print(df_scaled.corr())

# Step 2: Create the Covariance Matrix
cov_matrix = df_scaled.cov()

print("Covariance Matrix:")
print(cov_matrix.round(4))
```

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import KBinsDiscretizer

# Your exact input data
ages = np.array([15, 25, 35, 45, 55, 65]).reshape(-1, 1)

# 1. Uniform Strategy: Divides the range (15 to 65) into 3 equal-width intervals
est_uniform = KBinsDiscretizer(n_bins=3, encode='ordinal', strategy='uniform')
binned_uniform = est_uniform.fit_transform(ages)

# 2. Quantile Strategy: Divides data so each bin contains an equal number of samples (2 per bin)
est_quantile = KBinsDiscretizer(n_bins=3, encode='ordinal', strategy='quantile')
binned_quantile = est_quantile.fit_transform(ages)

# Combine into a clean DataFrame for comparison
df_result = pd.DataFrame({
    "Age": ages.flatten(),
    "Uniform_Bin": binned_uniform.flatten(),
    "Quantile_Bin": binned_quantile.flatten()
})

print(df_result)
print("\nUniform Bin Edges: ", est_uniform.bin_edges_[0])
print("Quantile Bin Edges:", est_quantile.bin_edges_[0])
```
